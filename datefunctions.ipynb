PySpark Date Functions

`python
from pyspark.sql.functions import *
`

Current Date & Time

| Function              | Syntax                                      | Description                 |
| --------------------- | ------------------------------------------- | --------------------------- |
| `current_date()`      | `df.withColumn("today", current_date())`    | Returns current system date |
| `current_timestamp()` | `df.withColumn("now", current_timestamp())` | Returns current timestamp   |
| `localtimestamp()`    | `df.withColumn("lt", localtimestamp())`     | Session local timestamp     |
| `now()`               | `df.withColumn("now", now())`               | Same as current_timestamp   |


Date Formatting & Conversion

| Function            | Syntax                                                                | Description           |
| ------------------- | --------------------------------------------------------------------- | --------------------- |
| `to_date()`         | `df.withColumn("dt", to_date("date_str","yyyy-MM-dd"))`               | String → Date         |
| `to_timestamp()`    | `df.withColumn("ts", to_timestamp("date_str","yyyy-MM-dd HH:mm:ss"))` | String → Timestamp    |
| `date_format()`     | `df.withColumn("fmt", date_format("dt","dd-MM-yyyy"))`                | Format date           |
| `unix_timestamp()`  | `df.withColumn("unix", unix_timestamp("date_str","yyyy-MM-dd"))`      | Converts to Unix time |
| `from_unixtime()`   | `df.withColumn("date", from_unixtime("unix"))`                        | Unix → Timestamp      |
| `cast("date")`      | `df.withColumn("dt", col("date_str").cast("date"))`                   | Type casting          |
| `cast("timestamp")` | `df.withColumn("ts", col("date_str").cast("timestamp"))`              | Type casting          |

Extract Date Parts

| Function       | Syntax                                    | Description    |
| -------------- | ----------------------------------------- | -------------- |
| `year()`       | `df.withColumn("year", year("dt"))`       | Extract year   |
| `month()`      | `df.withColumn("month", month("dt"))`     | Extract month  |
| `dayofmonth()` | `df.withColumn("day", dayofmonth("dt"))`  | Day of month   |
| `dayofyear()`  | `df.withColumn("doy", dayofyear("dt"))`   | Day of year    |
| `weekofyear()` | `df.withColumn("week", weekofyear("dt"))` | Week number    |
| `quarter()`    | `df.withColumn("qtr", quarter("dt"))`     | Quarter        |
| `hour()`       | `df.withColumn("hr", hour("ts"))`         | Extract hour   |
| `minute()`     | `df.withColumn("min", minute("ts"))`      | Extract minute |
| `second()`     | `df.withColumn("sec", second("ts"))`      | Extract second |

Date Arithmetic

| Function           | Syntax                                               | Description           |
| ------------------ | ---------------------------------------------------- | --------------------- |
| `date_add()`       | `df.withColumn("new", date_add("dt",10))`            | Add days              |
| `date_sub()`       | `df.withColumn("new", date_sub("dt",10))`            | Subtract days         |
| `datediff()`       | `df.withColumn("diff", datediff("end","start"))`     | Difference in days    |
| `add_months()`     | `df.withColumn("new", add_months("dt",2))`           | Add months            |
| `months_between()` | `df.withColumn("mb", months_between("end","start"))` | Months difference     |
| `next_day()`       | `df.withColumn("next", next_day("dt","Sunday"))`     | Next specific weekday |
| `last_day()`       | `df.withColumn("last", last_day("dt"))`              | Last day of month     |

Date Truncation & Rounding

| Function       | Syntax                                                 | Description        |
| -------------- | ------------------------------------------------------ | ------------------ |
| `trunc()`      | `df.withColumn("month_start", trunc("dt","month"))`    | Truncate to month  |
| `date_trunc()` | `df.withColumn("hour_trunc", date_trunc("hour","ts"))` | Truncate timestamp |
| `floor()`      | `df.select(floor(months_between("end","start")))`      | Floor value        |
| `ceil()`       | `df.select(ceil(months_between("end","start")))`       | Ceil value         |

Timestamp & Timezone

| Function               | Syntax                                                          | Description            |
| ---------------------- | --------------------------------------------------------------- | ---------------------- |
| `from_utc_timestamp()` | `df.withColumn("ist", from_utc_timestamp("ts","Asia/Kolkata"))` | Convert UTC → Timezone |
| `to_utc_timestamp()`   | `df.withColumn("utc", to_utc_timestamp("ts","Asia/Kolkata"))`   | Convert local → UTC    |

Common Exam Patterns (Very Important)

| Use Case                          | Syntax                                               |
| --------------------------------- | ---------------------------------------------------- |
| Calculate Age                     | `floor(months_between(current_date(),"dob")/12)`     |
| Last 7 days data                  | `df.filter(col("dt") >= date_sub(current_date(),7))` |
| Month start                       | `trunc("dt","month")`                                |
| Month end                         | `last_day("dt")`                                     |
| Difference between two timestamps | `datediff(to_date("ts1"), to_date("ts2"))`           |
| Convert string to proper date     | `to_date("col","yyyy-MM-dd")`                        |

Important Format Patterns

| Pattern | Meaning       |
| ------- | ------------- |
| yyyy    | 4-digit year  |
| yy      | 2-digit year  |
| MM      | Month (01–12) |
| dd      | Day           |
| HH      | Hour (24)     |
| mm      | Minute        |
| ss      | Second        |
