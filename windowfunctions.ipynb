PySpark Window Functions – One Table Cheat Sheet

| Category        | Function          | Syntax (PySpark DataFrame)                                                                    | Description                   |
| --------------- | ----------------- | --------------------------------------------------------------------------------------------- | ----------------------------- |
| **Window Spec** | Define Window     | `from pyspark.sql.window import Window`<br>`w = Window.partitionBy("dept").orderBy("salary")` | Creates window specification  |
|                 | Partition Only    | `Window.partitionBy("dept")`                                                                  | Divides data into groups      |
|                 | Order Only        | `Window.orderBy("salary")`                                                                    | Defines row ordering          |
|                 | Partition + Order | `Window.partitionBy("dept").orderBy("salary")`                                                | Most common window definition |
|                 | Rows Frame        | `w.rowsBetween(-1,1)`                                                                         | Row-based sliding window      |
|                 | Range Frame       | `w.rangeBetween(-1000,1000)`                                                                  | Value-based window            |


Ranking Functions

| Function         | Syntax                                                                                            | Description                      |
| ---------------- | ------------------------------------------------------------------------------------------------- | -------------------------------- |
| `row_number()`   | `from pyspark.sql.functions import row_number`<br>`df.withColumn("rn", row_number().over(w))`     | Unique sequential number         |
| `rank()`         | `from pyspark.sql.functions import rank`<br>`df.withColumn("rank", rank().over(w))`               | Same rank for ties, gaps allowed |
| `dense_rank()`   | `from pyspark.sql.functions import dense_rank`<br>`df.withColumn("drank", dense_rank().over(w))`  | Same rank, no gaps               |
| `percent_rank()` | `from pyspark.sql.functions import percent_rank`<br>`df.withColumn("pr", percent_rank().over(w))` | Relative rank (0–1)              |
| `ntile(n)`       | `from pyspark.sql.functions import ntile`<br>`df.withColumn("bucket", ntile(4).over(w))`          | Divides rows into n buckets      |

Analytical (Value) Functions

| Function           | Syntax                                                                                                 | Description             |
| ------------------ | ------------------------------------------------------------------------------------------------------ | ----------------------- |
| `lag()`            | `from pyspark.sql.functions import lag`<br>`df.withColumn("prev_sal", lag("salary",1).over(w))`        | Previous row value      |
| `lead()`           | `from pyspark.sql.functions import lead`<br>`df.withColumn("next_sal", lead("salary",1).over(w))`      | Next row value          |
| `first()`          | `from pyspark.sql.functions import first`<br>`df.withColumn("first_sal", first("salary").over(w))`     | First value in window   |
| `last()`           | `from pyspark.sql.functions import last`<br>`df.withColumn("last_sal", last("salary").over(w))`        | Last value in window    |
| `nth_value(col,n)` | `from pyspark.sql.functions import nth_value`<br>`df.withColumn("nth", nth_value("salary",2).over(w))` | Nth row value           |
| `cume_dist()`      | `from pyspark.sql.functions import cume_dist`<br>`df.withColumn("cd", cume_dist().over(w))`            | Cumulative distribution |


Aggregate Window Functions

| Function         | Syntax                                                                                                      | Description             |
| ---------------- | ----------------------------------------------------------------------------------------------------------- | ----------------------- |
| `sum()`          | `from pyspark.sql.functions import sum`<br>`df.withColumn("sum_sal", sum("salary").over(w))`                | Running sum             |
| `avg()`          | `from pyspark.sql.functions import avg`<br>`df.withColumn("avg_sal", avg("salary").over(w))`                | Running average         |
| `min()`          | `from pyspark.sql.functions import min`<br>`df.withColumn("min_sal", min("salary").over(w))`                | Running minimum         |
| `max()`          | `from pyspark.sql.functions import max`<br>`df.withColumn("max_sal", max("salary").over(w))`                | Running maximum         |
| `count()`        | `from pyspark.sql.functions import count`<br>`df.withColumn("cnt", count("salary").over(w))`                | Running count           |
| `collect_list()` | `from pyspark.sql.functions import collect_list`<br>`df.withColumn("list", collect_list("salary").over(w))` | Collects values as list |
| `collect_set()`  | `from pyspark.sql.functions import collect_set`<br>`df.withColumn("set", collect_set("salary").over(w))`    | Collects unique values  |


Common Exam Patterns

| Use Case                          | Window Spec                                                                                            |
| --------------------------------- | ------------------------------------------------------------------------------------------------------ |
| Running total                     | `Window.partitionBy("dept").orderBy("date").rowsBetween(Window.unboundedPreceding, Window.currentRow)` |
| Top N per group                   | Use `row_number()` + filter                                                                            |
| Remove duplicates (latest record) | `row_number().over(partitionBy("id").orderBy(desc("date")))`                                           |
| Compare current vs previous       | `lag()`                                                                                                |
| Salary difference                 | `col("salary") - lag("salary",1).over(w)`                                                              |
